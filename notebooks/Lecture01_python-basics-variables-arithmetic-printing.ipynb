## Types in materials data



Python types map onto materials data:

* `int` - counting: atomic number, number of specimens, number of grains
* `float` - measured: density (g/cm³), melting point (°C), yield strength (MPa)
* `str` - labels: crystal structure (`"fcc"`), phase (`"austenite"`), sample ID (`"A7-03"`)



*Are all strings the same kind of data?*

`"fcc"` is a **category** (small set of possible values).
`"Sample prepared at 1200°C for 2h in Ar atmosphere"` is **free text**.
Python treats them identically. They behave very differently in analysis.

*Does Python track units?*

No. `8.96` could be density in g/cm³ or length in meters. Good names
(`density_g_per_cm3`) help, but units are your responsibility.

# Arithmetic operators

| operator | meaning | example | result |
|---|---|---|---|
| `+` | addition | `4 + 3` | `7` |
| `-` | subtraction | `4 - 3` | `1` |
| `*` | multiplication | `4 * 3` | `12` |
| `/` | true division | `4 / 3` | `1.333...` |
| `//` | floor division | `4 // 3` | `1` |

## `**` vs `^`

Some operators use two characters. `**` means exponent, `//` means floor division, `!=` means not-equal (it means not-equal).

**Powers are `**`, never `^`:**

```python
print(2 ** 3)
```

```python
print(2 ^ 3)
```

`2 ** 3` → 8. `2 ^ 3` → 1 (bitwise XOR). **No error**, Python thinks you meant it.

## Order of operations

Same as algebra: parentheses first, then exponents, then `*`/`/`, then `+`/`-`.

When in doubt, add parentheses. They make the code clearer even when they are not strictly needed:

```python
mass_g = 44.5
volume_mL = 5.0
print(mass_g / volume_mL + 1)
print(mass_g / (volume_mL + 1))
```

Spaces around operators are optional but make code easier to read. `mass_g/volume_mL` and `mass_g / volume_mL` do the same thing. Convention: add spaces around `+`, `-`, `*`, `/`, `=`.

## Worked example: thermal expansion

* A 1 m steel rod: $L_0 = 1.000$ m, $\alpha = 17 \times 10^{-6}$ /°C (steel), $\Delta T = 80$ °C
* $\Delta L = L_0 \, \alpha \, \Delta T$

```python
L0 = 1.000
alpha = 17e-6
delta_T = 80
```

`17e-6` = scientific notation = $17 \times 10^{-6}$

```python
delta_L = L0 * alpha * delta_T
print(delta_L)
```

## Hands-on: theoretical density from the unit cell

(n, a, A, structure) = a minimal input record for this density model.
Four numbers and a label predict a measured property to &lt;1%.

The theoretical density of a crystal comes from its unit cell:

$$\rho = \frac{n A}{V_c N_A}$$

where $n$ is the number of atoms per unit cell, $A$ is the atomic weight (g/mol),
$V_c$ is the unit cell volume (cm³), and $N_A$ is Avogadro's number.

Copper is FCC:

| quantity | value |
|---|---|
| $n$ (atoms/cell) | 4 |
| $a$ (lattice parameter) | 0.36149 nm |
| $A$ (atomic weight) | 63.55 g/mol |
| $N_A$ | 6.022e23 /mol |
| handbook density | 8.96 g/cm³ |

The lattice parameter is in nm, but the density formula needs cm: 1 nm = 1e-7 cm.
