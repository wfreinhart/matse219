## Dimensionality

The **dimensionality** of a piece of data is how many indices it takes to look up a value:

| Dimensionality | What it looks like | Materials example |
|---|---|---|
| **Scalar** (0-D) | one number | density of a sample (2.70 g/cm³) |
| **Vector** (1-D) | a list of numbers | yield strength of 15 bolt specimens |
| **Matrix** (2-D) | rows and columns | stress-strain curve (two columns, many rows) |
| **Tensor** (3-D+) | matrix of matrices | a stack of 100 micrograph images (100 x 512 x 512) |

The general term for a data container of any dimensionality is an **array**. A vector is a
1-D array, a matrix is a 2-D array, and a tensor is a 3-D or higher array. NumPy
represents all of these with `numpy.ndarray`, created via `np.array()`.

A batch of 10 density readings stored in a single variable loses the individual
measurements. One number stored in an array adds complexity for no reason. The container
should match the data.

# Kinds of materials data

Materials data commonly include three kinds of information:

* **Counts** - how many: atoms per unit cell, number of specimens, valence electrons.
  Always whole numbers.
* **Measurements** - how much: density (g/cm³), melting point (°C), yield strength (MPa).
  Almost never land on a whole number.
* **Labels** - which one: crystal structure ("FCC"), phase ("austenite"), sample ID ("A7-03").
  Text, not numbers.

A fourth kind shows up often: **categories** vs. **free text**. "FCC" is one of a small
set of possible values. "Sample prepared at 1200°C for 2h in Ar atmosphere" is open-ended.
Both are text, but they behave very differently in analysis.

## Units

`8.96` could be density in g/cm³ or length in meters. Python does not know the difference.
Software enforces **syntax** (valid code), not **physical meaning** (correct science).
Keeping track of units is your responsibility.

```python
# density in g/cm^3
density_g_per_cm3 = 8.96

# convert to kg/m^3 (multiply by 1000)
density_kg_per_m3 = density_g_per_cm3 * 1000
print(density_kg_per_m3)
```

Descriptive variable names (`density_g_per_cm3`, not `x`) and comments keep track of units.
Libraries like Pint can automate unit-aware calculations in larger programs.

## `**` vs `^`

Some operators use two characters. `**` means exponent, `!=` means not-equal.

**Powers are `**`, never `^`:**

```python
print(2 ** 3)
```

```python
print(2 ^ 3)
```

`2 ** 3` → 8. `2 ^ 3` → 1 (bitwise XOR). **No error**, Python thinks you meant it.

## Order of operations

Same as algebra: parentheses first, then exponents, then `*`/`/`, then `+`/`-`.

When in doubt, add parentheses. They make the code clearer even when they are not strictly needed:

```python
mass_g = 44.5
volume_mL = 5.0
print(mass_g / volume_mL + 1)
print(mass_g / (volume_mL + 1))
```

Breaking a long calculation into named intermediate steps makes it easier to read and debug:

```python
mass_g = 44.5
volume_mL = 5.0
adjusted_volume_mL = volume_mL + 1
adjusted_density = mass_g / adjusted_volume_mL
print(adjusted_density)
```

Spaces around operators are optional but make code easier to read. `mass_g/volume_mL` and `mass_g / volume_mL` do the same thing. Convention: add spaces around `+`, `-`, `*`, `/`, `=`.

## Worked example: thermal expansion

* A 1 m steel rod: $L_0 = 1.000$ m, $\alpha = 17 \times 10^{-6}$ /°C (steel), $\Delta T = 80$ °C
* $\Delta L = L_0 \, \alpha \, \Delta T$

```python
L0 = 1.000
alpha = 17e-6
delta_T = 80
```

`17e-6` is Python's scientific notation for $17 \times 10^{-6}$

```python
delta_L = L0 * alpha * delta_T
print(delta_L)
```

## Hands-on: theoretical density from the unit cell

The theoretical density of a crystal comes from its unit cell:

$$\rho = \frac{n A}{V_c N_A}$$

where $n$ is the number of atoms per unit cell, $A$ is the atomic weight (g/mol),
$V_c$ is the unit cell volume (cm³), and $N_A$ is Avogadro's number.

<?xml version="1.0" encoding="utf-8"?>


Copper is FCC:

| quantity | value |
|---|---|
| $n$ (atoms/cell) | 4 |
| $a$ (lattice parameter) | 0.36149 nm |
| $A$ (atomic weight) | 63.55 g/mol |
| $N_A$ | 6.022e23 /mol |
| handbook density | 8.96 g/cm³ |

The lattice parameter is in nm, but the density formula needs cm: 1 nm = $10^{-7}$ cm.
