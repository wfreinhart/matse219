A rotating shaft may survive one loading cycle and still fail after the same cycle is repeated millions of times.

```python
#@title Load one S-N fatigue campaign (click ▶ to run, data loading, not a learning objective) { display-mode: "form" }
import os

import pandas as pd

_file = 'fatigue_42crmo4.csv'
_github = f'https://raw.githubusercontent.com/wfreinhart/matse219/main/datasets/{_file}'
_local = next(
    (p for p in (f'datasets/{_file}', f'../datasets/{_file}', f'../../datasets/{_file}',
                 f'../../../datasets/{_file}')
     if os.path.exists(p)),
    None,
)

try:
    _table = pd.read_csv(_local if _local else _github)
except Exception as e:
    raise RuntimeError(
        f"Could not load '{_file}'. If you are in Colab, check your internet "
        f"connection and that the file exists at {_github}"
    ) from e

_curve = _table[_table['curve_id'] == 29].sort_values('stress_amplitude_MPa', ascending=False)

stress_MPa = _curve['stress_amplitude_MPa'].tolist()
cycles_completed = _curve['cycles'].tolist()
is_runout = _curve['is_runout'].tolist()

print(f'Loaded one fatigue campaign on 42CrMo4 steel: {len(stress_MPa)} specimens')
print(f'  stress amplitude: {min(stress_MPa):.0f} to {max(stress_MPa):.0f} MPa')
print(f'  cycles completed: {min(cycles_completed)} to {max(cycles_completed)}')
print(f'  runouts: {sum(is_runout)} specimen(s) stopped before breaking')
```
