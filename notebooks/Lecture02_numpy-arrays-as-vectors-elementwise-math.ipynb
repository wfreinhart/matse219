## Dimensionality in materials data



Materials data comes at different scales of complexity:

* **Scalar** - a single number (atomic number of iron: `26`, density of copper: `8.96`)
* **Vector** - an ordered sequence (the `mass` array above holds 10 measurements from one experiment)


*What does the data from a test like this actually look like?*

Here are 12 points from the elastic region of a published stress-strain curve
([Zheng et al., Science 344, 2014](https://doi.org/10.1126/science.1252291)):

````python
strain = np.array([0.002, 0.004, 0.006, 0.008, 0.010, 0.012, 0.014, 0.016, 0.018, 0.020, 0.022, 0.024])
stress_MPa = np.array([1.2, 2.5, 3.7, 5.0, 6.2, 7.4, 8.7, 10.0, 11.2, 12.4, 13.7, 15.0])
print(strain)
print(stress_MPa)
````

Two parallel arrays, 12 entries each. The elastic modulus is the slope:

````python
modulus = stress_MPa / strain
print(modulus)
print(modulus)
````

One line of elementwise division, 12 modulus estimates.

# Elementwise arithmetic

*All 10 densities in one line:*

````python
density = mass / volume
print(density)
````

````python
print(type(density))
print(len(density))
````

NumPy divided entry by entry: first mass with first volume, second with second, and so on.

The two arrays are **parallel**: index i refers to the same specimen in both.
Shuffled order? Division still runs, densities come out wrong. NumPy rejects
incompatible lengths but cannot check specimen order, units, or physical meaning.

## Broadcasting: adding a scalar to every entry

````python
print(mass + 0.5)
````

Same as $\vec{m} + 0.5 = (m_1 + 0.5,\ m_2 + 0.5,\ \ldots)$

Works with any operator:

````python
print(mass - 2.0)
print((mass + 0.5) / volume)
````

## Combining two arrays

````python
print(mass + volume)
print(mass * volume)
````

Entry-by-entry: first mass with first volume, second with second, and so on.

NumPy checks shapes, not units. `mass + volume` (g + cm³) is physically meaningless but runs fine.

# Guided practice: screening a batch

Pure aluminum: about 2.70 g/cm³. Values far from that suggest a mislabeled
sample or measurement error.

````python
print(np.min(density), np.max(density))
````

Range: 2.60 to 2.83 g/cm³, all plausible for aluminum alloys. Quick batch screening
like this is one of the first things you do with lab data.

Spot-check: sample 0 should be 27.0 / 10.0 = 2.7.

`[0]` grabs the first entry, `[1]` the second, and so on.

````python
print(mass[0], volume[0], density[0])
````

## Hands-on: are lead atoms really big?

Pure metals span a wide range of density. Below are 14 common structural metals
with their atomic mass and handbook density at room temperature.

````python
metals = np.array(['Cu','Ni','Al','Zn','Pb','Sc','Ti','V','Cr','Ga','Sn','Co','In','Fe'])
atomic_mass = np.array([63.546, 58.693, 26.981, 65.38, 207.2, 44.955, 47.867, 50.942, 51.996, 69.723, 118.71, 58.933, 114.818, 55.845])
density_handbook = np.array([8.96, 8.908, 2.70, 7.14, 11.34, 2.985, 4.506, 6.11, 7.19, 5.91, 7.265, 8.90, 7.31, 7.874])

print(metals)
print(atomic_mass)
print(density_handbook)
````

Lead's density, 11.34 g/cm³, is by far the highest in this list. Does that mean a
lead atom takes up more space than the others?

Molar volume answers that directly: the volume occupied by one mole of atoms,
$V_m = A / \rho$ (cm³/mol). In the code cell below, compute `molar_volume` from
`atomic_mass` and `density_handbook`, then print it.

Use `np.min` and `np.max` to find the smallest and largest molar volume. Compare
the ratio (max / min) for `density_handbook` to the ratio for `molar_volume`.

Density spans about 4.2x across these 14 metals, but molar volume spans only about
2.8x. Lead's high density comes mostly from its large atomic mass, 207.2 g/mol,
more than 3x copper's, not from unusually large atoms.
