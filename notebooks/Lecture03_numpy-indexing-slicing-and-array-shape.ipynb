# Setup: 20 elements

The periodic table: one of the oldest datasets in materials science.

Every element has dozens of measured properties. Today: indexing and
slicing on the first 20.

```python
import numpy as np

elements = np.array(['H', 'He', 'Li', 'Be', 'B', 'C', 'N', 'O', 'F', 'Ne',
                      'Na', 'Mg', 'Al', 'Si', 'P', 'S', 'Cl', 'Ar', 'K', 'Ca'])
melting_K = np.array([14.01, 0.95, 453.65, 1560, 2349, 3823, 63.15, 54.36,
                       53.53, 24.56, 370.87, 923, 933.47, 1687, 317.3,
                       388.36, 171.6, 83.80, 336.53, 1115])
print(elements)
print(melting_K)
```

These arrays are **parallel**: `elements[i]` and `melting_K[i]` describe the same element.

```python
print(f'index 0: {elements[0]}  ({melting_K[0]:.2f} K)')
print(f'index 1: {elements[1]}  ({melting_K[1]:.2f} K)')
print(f'index 2: {elements[2]}  ({melting_K[2]:.2f} K)')
print(f'index 3: {elements[3]}  ({melting_K[3]:.2f} K)')
```

# Zero-based indexing

*What is carbon's melting point?*

Carbon is the 6th element listed → index **5** (counting starts at 0):

```python
print(melting_K[6])   # this is nitrogen, not carbon!
```

```python
print(elements[0])
print(melting_K[5])
print(elements[5])
```

Rule: `array[i]` = the `(i+1)`-th item. Valid indices: `0` to `len-1`.

# Slicing: a range of entries

`array[start:stop]` grabs from `start` up to but **not including** `stop`:

Period 2 (Li through Ne) and period 3 (Na through Ar):

```python
print(elements[2:10])
print(melting_K[2:10])
```

```python
print(elements[10:18])
print(melting_K[10:18])
```

Melting points rise toward strongly bonded elements (C: 3823 K, Si: 1687 K),
then drop at the noble gases. Not perfectly smooth, but the pattern repeats.
Period boundaries make better slices than arbitrary splits like `[0:10]`.

Omit `start` or `stop` to mean "from the beginning" or "to the end":

```python
print(elements[:5])
print(elements[15:])
```

With a **step**: `array[start:stop:step]`

```python
print(elements[::2])     # every other, starting from 0
print(elements[1::2])    # every other, starting from 1
```

```python
print(elements[-5:])     # last five
```

# Fancy indexing: arbitrary positions

The alkali metals Li, Na, K are at indices 2, 10, 18, not a contiguous range.
Use a **collection of indices** inside square brackets to select several positions at once:

```python
alkali_idx = [2, 10, 18]
print(elements[alkali_idx])
print(melting_K[alkali_idx])
```

454, 371, 337 K: melting point decreases down the group.

*Why?* Atomic radius increases (Li < Na < K), so the metallic bond weakens.
Students in intro MatSE see this same trend. Here it falls out of a fancy-index
selection: a chemical family in the table is a noncontiguous set of array indices.

The index spacing (2, 10, 18 differ by 8) reflects period lengths. This array is
ordered by atomic number, so index = atomic number - 1. Position encodes identity
only because the ordering convention is preserved.

Fancy indexing is a NumPy feature. It does not work on every kind of Python container.

> **Read more:** [PDSH — Fancy Indexing](https://jakevdp.github.io/PythonDataScienceHandbook/02.07-fancy-indexing.html)

# 2D arrays

*What if each element has multiple properties?*

Stack two 1-D arrays into one 2-D array:

```python
radius_pm = np.array([25, 31, 145, 105, 85, 70, 65, 60, 50, 38,
                       180, 150, 125, 110, 100, 100, 100, 71, 220, 180])
properties = np.array([melting_K, radius_pm])
print(properties)
```

```python
print(properties.shape)
```

`(2, 20)` = 2 rows (properties), 20 columns (elements).

**Dimensionality:** a **matrix**, 2-D, rows x columns.

* L01: single numbers (scalars)
* L02: 1-D arrays of measurements (vectors)
* Now: a 2-D property table, two properties on 20 elements

Real materials datasets are exactly this shape: rows are properties or specimens,
columns are elements or measurements.

## 2D indexing: `array[row, col]`

```python
print(properties[0, 5])    # row 0 (melting pt), col 5 (carbon)
print(properties[1, 5])    # row 1 (radius), col 5 (carbon)
```

Grab an entire row or column with `:`:

```python
print(properties[0, :])    # all melting points
print(properties[:, 5])    # both properties for carbon
```

## `axis` preview

`np.mean` computes an average. With a 2-D array,
you choose which direction to average over:

* `axis=0` - down the rows (across properties, for one element)
* `axis=1` - across columns (across elements, for one property)

```python
print(np.mean(properties, axis=1))   # 2 values: avg melting pt, avg radius
```

```python
print(np.mean(properties, axis=1).shape)   # 2 values (one per property)
print(np.mean(properties, axis=0).shape)   # 20 values (one per element)
```

`axis=1`: one average per property row (K stays K, pm stays pm).
`axis=0`: averages K with pm for each element, physically meaningless.
The shape permits either axis. Units determine which makes sense.

The same 14 metals from last lecture, this time with melting point and Young's
