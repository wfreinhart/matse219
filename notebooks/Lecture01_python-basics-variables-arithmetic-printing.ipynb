## `**` vs `^`

Some operators use two characters. `**` means exponent, `//` means floor division, `!=` means not-equal (it means not-equal).

**Powers are `**`, never `^`:**

```python
print(2 ** 3)
```

```python
print(2 ^ 3)
```

`2 ** 3` → 8. `2 ^ 3` → 1 (bitwise XOR). **No error**, Python thinks you meant it.

## Order of operations

Same as algebra: parentheses first, then exponents, then `*`/`/`, then `+`/`-`.

When in doubt, add parentheses. They make the code clearer even when they are not strictly needed:

```python
mass_g = 44.5
volume_mL = 5.0
print(mass_g / volume_mL + 1)
print(mass_g / (volume_mL + 1))
```

Spaces around operators are optional but make code easier to read. `mass_g/volume_mL` and `mass_g / volume_mL` do the same thing. Convention: add spaces around `+`, `-`, `*`, `/`, `=`.

$$\rho = \frac{n A}{V_c N_A}$$
