# Not everything is a number

`arrays` require one dtype for everything inside.

*What happens if we try to sneak in something that isn't a number?*

```python
import numpy as np

my_array = np.array(['Fe', 26, 7.87])
print(my_array)
print(my_array.dtype)
```

Silent conversion — everything became a `string` (`dtype` starts with `<U`). No error, no warning.

Need name + symbol + atomic number + density together → mixed types. Two built-in containers solve this: `list` and `dict`.

# Lists: ordered, mixed, and changeable

Square brackets, like `array` — but no type restriction.

```python
symbols = ['H', 'He', 'Li', 'Be']
print(symbols)
print(type(symbols))
```

## Indexing and slicing (same syntax as arrays)

```python
print(symbols[0])
print(symbols[-1])
print(symbols[1:3])
print(symbols[::-1])
```

## Mixing types

*Can one list hold a string, an int, and a float side by side?*

```python
iron_row = ['Iron', 'Fe', 26, 55.85, 7.87]
print(iron_row)
for item in iron_row:
    print(item, type(item))
```

`str`, `str`, `int`, `float`, `float` — nothing upcast.

## Lists can change after you make them

`arrays`: fixed size. `lists`: **mutable**.

```python
symbols = ['H', 'He', 'Li', 'Be']
symbols.append('B')           # add an entry at the end
print(symbols)

symbols[0] = 'H*'             # overwrite an entry by position
print(symbols)

symbols.remove('He')          # delete an entry by value
print(symbols)
```

`len()` — same as `arrays`:

```python
print(len(symbols))
```

# The `+1` reversal: what arrays gave us for free

Densities (g/cm$^3$) of four metals, in order Fe, Cu, Al, Au:

```python
densities_list = [7.87, 8.96, 2.70, 19.3]
print(densities_list)
```

*Add 1 to every entry, the way we did with arrays?*

```python
# EXPECTED-ERROR: this cell fails on purpose -- see the surrounding text
densities_list + 1
```

`TypeError` — a `list` isn't a vector to Python.

Multiply instead, hoping to double every density:

```python
print(densities_list * 2)
```

No error — but `*` on a `list` means "repeat the sequence," not "scale each element." Same four numbers, twice, concatenated.

```python
densities_array = np.array(densities_list)
print(densities_array + 1)   # elementwise addition
print(densities_array * 2)   # elementwise multiplication
```

`lists` are for flexible containers; `arrays` are for flexible math.

# Dictionaries: labels instead of positions

*Was density index 3 or 4 in `iron_row`?* `dicts` look things up by **label** ("key") instead of position.

```python
iron = {}
iron['name'] = 'Iron'
iron['symbol'] = 'Fe'
iron['density'] = 7.87
iron['melting_point'] = 1538
print(iron)
```

Same thing, all at once — `{key: value, ...}`:

```python
iron = {'name': 'Iron', 'symbol': 'Fe', 'density': 7.87, 'melting_point': 1538}
print(iron)
```

## Accessing values by key

```python
print(iron['density'])
print(iron['name'], 'has a melting point of', iron['melting_point'], 'degC')
```

## `KeyError`

*What happens if we ask for a key that doesn't exist?*

```python
# EXPECTED-ERROR: this cell fails on purpose -- see the surrounding text
print(iron['boiling_point'])
```

`dict` version of `IndexError` — same mistake, different name.

`.get()` supplies a fallback instead of crashing:

```python
print(iron.get('boiling_point', 'unknown'))
print(iron.get('density', 'unknown'))
```

## Adding and updating keys

New key → added. Existing key → overwritten. Same syntax:

```python
iron['boiling_point'] = 2862   # add a new key
print(iron)

iron['density'] = 7.874        # update an existing key with a more precise value
print(iron)
```

## Looking at everything in a `dict`

```python
print(iron.keys())
print(iron.values())
print(iron.items())
```

`.items()` + `for` loop (loops properly covered soon) — a preview:

```python
for key, value in iron.items():
    print(key, '->', value)
```

A second dict, same keys as `iron` — we'll need it in a minute:

```python
copper = {'name': 'Copper', 'symbol': 'Cu', 'density': 8.96, 'melting_point': 1085}
print(copper)
```

# Dict-of-lists: the shape of a table

*Four elements, four separate dicts — what about 50?* Flip the structure: one `dict`, each key maps to a **list** of values, one entry per element.

```python
elements = {
    'symbol':        ['Fe',  'Cu',  'Al',  'Au'],
    'name':          ['Iron', 'Copper', 'Aluminum', 'Gold'],
    'density':       [7.87, 8.96, 2.70, 19.3],
    'melting_point': [1538, 1085, 660, 1064],
}
print(elements)
```

## Two-step lookup: column, then cell

```python
print(elements['density'])
```

Key first, then position — "density of the third element":

```python
print(elements['density'][2])
```

*How would you get all of Iron's properties back at once — its whole row?* No clean way yet — you'd pull `elements['symbol'][0]`, `elements['name'][0]`, ... one at a time.

### [Check your understanding]

Silver (`Ag`) has a density of 10.49 g/cm$^3$ and a melting point of 962 degC.
Add it to `elements` by calling `.append()` on all four lists (one line per column), then print `elements` to confirm it worked.

```python
elements['symbol'].append('Ag')
elements['name'].append('Silver')
elements['density'].append(10.49)
elements['melting_point'].append(962)
print(elements)
```

All four `.append()` calls had to happen together — miss one and the columns fall out of sync.

# Seeing `elements` as a table

One row at a time, using the shared position across all four lists:

```python
for i in range(len(elements['symbol'])):
    print(f"{elements['symbol'][i]:>3}  {elements['name'][i]:<10} "
          f"{elements['density'][i]:>6.2f} g/cm^3  {elements['melting_point'][i]:>5} degC")
```
