```python
#@title Load one S-N fatigue campaign (click ▶ to run, data loading, not a learning objective) { display-mode: "form" }
import os

import pandas as pd

_file = 'fatigue_42crmo4.csv'
_github = f'https://raw.githubusercontent.com/wfreinhart/matse219/main/datasets/{_file}'
_local = next(
    (p for p in (f'datasets/{_file}', f'../datasets/{_file}', f'../../datasets/{_file}',
                 f'../../../datasets/{_file}')
     if os.path.exists(p)),
    None,
)

try:
    _table = pd.read_csv(_local if _local else _github)
except Exception as e:
    raise RuntimeError(
        f"Could not load '{_file}'. If you are in Colab, check your internet "
        f"connection and that the file exists at {_github}"
    ) from e

_curve = _table[_table['curve_id'] == 29].sort_values('stress_amplitude_MPa', ascending=False)

stress = _curve['stress_amplitude_MPa'].tolist()
cycles = _curve['cycles'].tolist()
runout = _curve['is_runout'].tolist()

print(f'Loaded one fatigue campaign on 42CrMo4 steel: {len(stress)} specimens')
print(f'  stress: {min(stress):.0f} to {max(stress):.0f} MPa')
print(f'  cycles: {min(cycles)} to {max(cycles)}')
print(f'  runout: {sum(runout)} specimen(s) stopped before breaking')
```
