You have three parallel lists.

- `stress` is the amplitude each specimen was cycled at, in MPa

- `runout` is `True` for a specimen that had not broken when the test was stopped

# Wrap-up

- A `for` loop runs its indented block once per entry, and a line after the block runs once.
- `range(len(...))` gives the position, which keeps parallel lists aligned.
- A counter starts at zero before the loop and grows only when an `if` says so.
- In an `if` / `elif` / `else` chain only the first true branch runs, so the order of the tests matters.
