## `.iloc` and `.loc`

`.iloc` selects by integer position, and `.loc` selects by label.
`data.loc[rows, columns]` takes the rows before the comma and the columns after it.

```
first_row = data.__[0]                       # .iloc: position 0, the first row
print(first_row['eid'])

first_three = data.iloc[__]                  # .iloc: positions 0 up to (not including) 3
print(first_three.shape)

by_label = data.__[0:2, ['eid', 'YS_MPa']]   # .loc: label slice 0 through 2, plus two columns
by_label
```

## Comparisons, masks, and counting

A comparison against a column gives one `True`/`False` answer per row: a mask.
`np.sum` counts the `True` values, and `np.mean` gives their fraction.

```
is_tension = data['test_mode'] __ 'tension'   # == : True where the test mode is tension
print(is_tension.dtype)                        # bool, so this really is a mask

n_tension = np.__(is_tension)      # np.sum: count of True values
frac_tension = np.__(is_tension)   # np.mean: fraction of rows that are True
print(f'{n_tension} of {len(data)} rows are tension tests ({frac_tension:.1%})')
```

## Filtering with `.loc[mask]`

Put the mask in the row position of `.loc` to keep only the rows marked `True`.

```
strong = data['YS_MPa'] > 800   # mask: yield strength above 800 MPa
strong_alloys = data.__[__]     # keep only the rows where the mask is True

print(strong_alloys.shape)
strong_alloys[['eid', 'YS_MPa']].head()
```

## Missing values: `.dropna(subset=[...])`

`subset` names the columns that must be present for a row to stay.
A row stays even if some other column is missing.

```
both = data.drop__(subset=['YS_MPa', 'UTS_MPa'])   # keep rows with both measurements

print(f'{len(both)} of {len(data)} alloys report both properties')
both[['eid', 'YS_MPa', 'UTS_MPa']].head()
```

## Descriptive statistics with NumPy functions

Compute the mean, median, and standard deviation of the clean yield-strength column above.

```
ys = both['YS_MPa']    # the clean yield-strength column from the previous example

mean_ys = np.__(ys)     # np.mean: the average
median_ys = np.__(ys)   # np.median: the middle value
std_ys = np.__(ys)      # np.std: how spread out the values are

print(f'mean:   {mean_ys:.0f} MPa')
print(f'median: {median_ys:.0f} MPa')
print(f'std:    {std_ys:.0f} MPa')
```

## Percentiles and the interquartile range

The interquartile range is the width of the middle half of the data: Q3 minus Q1.

```
uts = both['UTS_MPa']    # the clean tensile-strength column

q1 = np.percentile(uts, __)   # 25th percentile
q3 = np.percentile(uts, __)   # 75th percentile
iqr = __ - __                 # interquartile range

print(f'Q1 = {q1:.0f} MPa, Q3 = {q3:.0f} MPa')
print(f'IQR = {iqr:.0f} MPa')
```

## Plotting a function: `np.linspace` and `ax.plot`

Plot $y = x^2$ using 51 evenly spaced x values from 0 to 5.

## Histograms and bar charts: `ax.hist` and `ax.bar`

Make a histogram of hardness, and a bar chart comparing mean yield strength by test mode.

```
hv = data['HV'].dropna()   # hardness values, with the missing ones removed

tension_ys = data.loc[data['test_mode'] == 'tension', 'YS_MPa'].dropna()
compression_ys = data.loc[data['test_mode'] == 'compression', 'YS_MPa'].dropna()

fig, axes = plt.subplots(1, 2, figsize=(9, 4))

axes[0].__(hv)                       # ax.hist: counts of hardness values by bin
axes[0].set_xlabel('Vickers hardness')
axes[0].set_ylabel('count')

names = ['tension', 'compression']
means = [np.mean(tension_ys), np.mean(compression_ys)]
axes[1].__(names, means)             # ax.bar: one bar per category
axes[1].set_ylabel('mean yield strength (MPa)')

fig.tight_layout()
```

## Correlation: `stats.pearsonr`

Compute the Pearson `r` between yield strength and ultimate tensile strength.

```
x = both['YS_MPa']
y = both['UTS_MPa']

r, p = stats.__(x, y)   # stats.pearsonr: returns r and a p-value, unpacked like fig, ax
print(f'r = {__:.3f}')    # r: the correlation coefficient itself
```

## Fitting a line: `stats.linregress`

Fit a linear model for ultimate tensile strength as a function of yield strength.
Print it in the form `y = mx + b` with the actual values of `m` and `b`.

```
fit = stats.__(x, y)   # fit a straight line by least squares

print(f'slope     = {fit.__:.3f}')   # the line's slope
print(f'intercept = {fit.__:.1f} MPa')   # the line's intercept
print(f'y = {fit.slope:.3f} x + {fit.intercept:.1f}')
```

## Residuals and R-squared

Compute the residuals for the model above, and report their spread and the R-squared.

```
y_hat = fit.slope * x + fit.__   # predicted UTS for every alloy

residuals = y - y_hat                # measured minus predicted
mean_residuals = y - np.mean(y)      # what the mean-only model gets wrong

r2 = 1 - np.var(__) / np.var(mean_residuals)   # how much of the spread the line removes
print(f'residual standard deviation = {np.std(residuals, __=1):.1f} MPa')
print(f'R^2 = {r2:.3f}')
```
